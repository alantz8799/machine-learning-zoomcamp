```python
df_test = df[df.origin == 'Europe'][['vehicle_weight', 'model_year']][:7]
df_test
```

```python
X = df_test.values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = (XTX_inv @ X.T) @ y
w
```
